```python
%load_ext autoreload
%autoreload 2

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import (
    accuracy_score,
    confusion_matrix,
    classification_report)

from sklearn_nominal import ZeroRClassifier
from sklearn_nominal import OneRClassifier
from sklearn_nominal import PRISMClassifier
from sklearn_nominal import CN2Classifier
```

```python
import warnings
warnings.filterwarnings("ignore")
```

```python
PATH='https://weblidi.info.unlp.edu.ar/catedras/MD_SI/DATOS/'
df = pd.read_csv(PATH + 'Drug5.csv')
#df = pd.read_csv('Drug5_atipicos.csv')
df.head()

X = df.drop('Drug', axis=1)
y = df['Drug']
```

```python
df.info()
```

```python
def validacion(y, y_pred):
    print("\nMETRICAS")
    print("-"*20)

    # Accuracy
    score = accuracy_score(y, y_pred)
    print(f"accuracy : {score:.2f}")

    # Matriz de Confusión
    cm = confusion_matrix(y, y_pred)
    print("\nMatriz de confusión")
    print(cm)

    # Métricas de Clasificación
    print('\nReporte de Clasificación:')
    print(classification_report(y, y_pred))
```

```python
#-- Atributos nominales y numéricos ---
print(X.head())

model = OneRClassifier(criterion="accuracy")
model.fit(X, y)

y_pred = model.predict(X)

print("\nREGLAS - OneR")
print("-"*20)
print(model.pretty_print(np.unique(y)))

validacion(y, y_pred)
```

```python
#-- Atributos numéricos --
X_num = pd.get_dummies(X)
print(X_num.head())

model = OneRClassifier()
model.fit(X_num, y)

y_pred = model.predict(X_num)

print("\nREGLAS - OneR")
print("-"*20)
print(model.pretty_print(np.unique(y)))

validacion(y, y_pred)
```
